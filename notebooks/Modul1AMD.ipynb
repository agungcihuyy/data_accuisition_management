# Pentingnya Reproducibility dalam Akuisisi Data dan contoh kasus nyatanya
Reproducibility penting dalam akuisisi data karena hasil analisis hanya dapat dipercaya jika orang lain bisa mengulang langkah pengambilan data dan memperoleh hasil yang sama. Tanpa jejak yang jelas, mulai dari sumber data, waktu pengambilan, hingga skrip yang dipakai, kesalahan sulit dilacak dan kesimpulan tidak dapat diverifikasi. Reproducibility juga melindungi data mentah dari penimpaan, memudahkan kolaborasi dalam tim, dan menghemat waktu ketika data perlu diperbarui di kemudian hari. Contoh nyata adalah makalah ekonom Reinhart dan Rogoff tahun 2010 yang menyimpulkan bahwa utang publik di atas 90 persen PDB menekan pertumbuhan ekonomi. Temuan ini banyak dikutip dalam perdebatan kebijakan penghematan anggaran di berbagai negara. Pada 2013, Thomas Herndon, mahasiswa doktoral di University of Massachusetts Amherst, berhasil mengulang analisis tersebut dan menemukan kesalahan pada spreadsheet, yaitu beberapa negara tidak ikut terhitung dalam rata-rata. Setelah dikoreksi, penurunan pertumbuhan yang dilaporkan menjadi jauh lebih kecil dan kesimpulan awalnya melemah. Kasus ini menunjukkan bahwa data dan langkah analisis yang terdokumentasi memungkinkan kesalahan ditemukan oleh pihak lain. Karena itu, praktikum ini mewajibkan notebook, Git, dan struktur folder yang konsisten agar setiap langkah akuisisi dapat ditelusuri dan diulang.
